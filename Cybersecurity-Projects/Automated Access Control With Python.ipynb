## Project Overview: Automated Access Control Verification with Python
In access and identity governance, relying on manual verification slows down operations and opens the door to human error. This project implements a lightweight Python automation workflow to evaluate access requests against two synchronized security checks before granting entry:

1. **User Identity Validation:** Confirms whether an individual holds active authorization in the system directory.
2. **Hardware Asset Attribution:** Verifies that the physical device presented matches the hardware uniquely registered to that person.

By tying user clearance directly to specific hardware tags, the script eliminates guesswork at security checkpoints, speeds up turnaround for IT staff, and enforces a strict fail-closed, least-privilege access model.

## Lead Security Analyst: Matthew Abegyah


### Operational Scenario

Physical security checkpoints and IT service desks need fast, reliable ways to validate credentials and equipment. When employees bring company hardware to the desk, an analyst must verify two critical details before approving network provisioning, maintenance access, or physical sign-off:

1. **System Authorization Status:** Does the employee currently hold active system clearance?
2. **Device-to-User Binding:** Does the hardware asset tag match the specific serial or device ID registered to that employee in inventory?

Checking disparate spreadsheets by hand slows down service queues, introduces human error, and risks allowing unauthorized or swapped devices onto internal networks.

### Technical Objectives

1. **Automate Verification:** Replace manual lookups with a fast, reliable Python routine.
2. **Prevent Asset Drift:** Catch unapproved hardware swaps and flag unauthorized individuals presenting corporate gear.
3. **Enforce Fail-Closed Access:** Default to denying access unless both the identity and the device serial match approved records.

This task encapsulates the entire verification algorithm into a reusable, parameterized function named login(). It utilizes nested conditional statements to enforce a hierarchical authorization check: identity validation occurs first, followed by device binding verification.

```python
# Assign `approved_users` to a list of approved usernames

approved_users = ["elarson", "bmoreno", "sgilmore", "eraab", "gesparza"]

# Assign `approved_devices` to a list of device IDs that correspond to the usernames in `approved_users`

approved_devices = ["8rp2k75", "hl0s5o1", "4n482ts", "a307vir", "3rcv4w6"]

# Define a function named `login` that takes in two parameters, `username` and `device_id`

def login(username, device_id):

    # If `username` belongs to `approved_users`, 

    if username in approved_users:

        # then display "The user ______ is approved to access the system.",

        print("The user", username, "is approved to access the system.")

        # assign `ind` to the index of `username` in `approved_users`,

        ind = approved_users.index(username)

        # and execute the following conditional
        # If `device_id` matches the element at the index `ind` in `approved_devices`,

        if device_id == approved_devices[ind]:

          # Display a message if a user is asigned to the right device_" 

          print(device_id, "is the assigned device for", username)

        #Check and diaplay a message if approved user in the approved list is matched with the wrong device

        else:
            print(device_id, "is not their assigned device.")
  
    # Otherwise (part of the outer conditional and handles the case when `username` does not belong to `approved_users`),

    else:

        # Display a message when a user in not in the approved_list"

        print("The username", username, "is not approved to access the system.")

# Call the function you just defined to experiment with different username and device_id combinations

login("bmoreno", "hl0s5o1")
login("elarson", "r2s5r9g")
login("abernard", "4n482ts")
```
